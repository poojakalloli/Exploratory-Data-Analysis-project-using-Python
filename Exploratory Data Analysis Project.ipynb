```python
df['year_introduced'].value_counts()
```

```python
ax = df['year_introduced'].value_counts() \
      .head(10) \
      .plot(kind='bar', title="Top 10 Years Coasters Introduced")

ax.set_xlabel('year_introduced')
ax.set_ylabel("Count")
```

```python
ax=df['speed_mph'].plot(kind='hist',
                        bins=20,
                        title='Coaster Speed(mph)')
ax.set_xlabel('speed_mph')
```

```python
ax=df['speed_mph'].plot(kind='kde',
                
                        title='Coaster Speed(mph)')
ax.set_xlabel('speed_mph')
```

```python
df.plot(kind='scatter',
        x='speed_mph',
        y='height_ft',
        title='Coaster Speed(mph)vs hegiht')
plt.show()
```

```python
ax=sns.scatterplot(x='speed_mph',
                   y='height_ft',
                   hue='year_introduced',
                   data=df)
ax.set_title('Coaster speed vs height')
plt.show()
```

```python
df_corr=df[['year_introduced','speed_mph',
            'height_ft']].dropna().corr()
df_corr
```

```python
sns.heatmap(df_corr,annot=True)
```

```python
df = df[['Year_Introduced','Speed_mph','Height_ft','Inversions_clean','Gforce_clean','Type_Main']]

# Remove missing values
df = df.dropna()

# Create pairplot
sns.pairplot(df, hue='Type_Main')

# Show plot
plt.show()
```
